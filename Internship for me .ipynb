# ARGUMENTS

```python
def find_avg(*number):
    sum=0
    for i in number:
        sum+=i
    print(sum/(len(number)))
find_avg(2,3)

    
```

```python
def test_var(f_arg, *argv):
    print("First Argument is:" , f_arg)
    for arg in argv:
        print("Another arg is: ", arg)
test_var("Mohamed", "Samy", "Omar", "Sameh")
    
```

# Kw Argument

```python
def show_details(a,b,*args,**kw):
     print("a is : ",a)
     print("b is : ",b)
     print("*args is :", args)
     print("**kw is :", kw)
show_details(1,2,3,4,5,6,c=7,d=8,e=9)
```
