# معرفة كل الاستايلات المتاخة

```python
import matplotlib.pyplot as plt

print(plt.style.available)
```

```python
plt.style.use('classic')

Sales = (1, 2, 3, 6, 8, 5, 4,7)
Revenue= (1, 3, 7, 4, 3, 2, 4,5)
Returns=(2,3,5,4,7,6,8,7)
x= (1,2,3,4,5,6,7,8)

plt.plot(x, Sales, color='red', marker='o',markersize=10,linestyle='solid', linewidth=2,label='Sales')
plt.plot(x,Revenue , color='Green', marker='o',markersize=10,linestyle='dotted', linewidth=2,label='Monthly Date')
plt.plot(x, Returns, color='Blue', marker='o',markersize=10,linestyle='dashdot', linewidth=2,label='Returns')
plt.legend()

# solid - dashed - dashdot- dotted  
plt.ylabel('Sales')
plt.xlabel('Monthly Date')
plt.title('Relation Between Hour Work And Salary')
plt.show()
```

```python
plt.style.use('fivethirtyeight')

a = (1, 2, 3, 4, 5, 6, 7)
b = (1, 2, 3, 8, 5, 6, 7)

plt.plot(b, a)
plt.show()
```

```python
plt.plot(b,a)
plt.savefig("graph.png",dpi=300)
plt.show()
```

# Pie

```python
import matplotlib.pyplot as plt

plt.style.use('classic')

y = [20, 40, 10, 20, 10]

plt.pie(
    y,
    labels=['Area1', 'Area2', 'Area3', 'Area4', 'Area5'],
    explode=[0.1, 0.2, 0.1, 0.1, 0.1],
    autopct='%1.1f%%',
    shadow=True
)

plt.title('Area Sales')
plt.axis('equal')
plt.show()



```

# Bar

```python
x= (1,2,3,4,5,6,7,8)
Sales = (1, 2, 3, 6, 8, 5, 4,7)
plt.bar(x,Sales)
plt.text(4,-1,'Sales per Area',bbox=({'facecolor':'orange','alpha':0.9}))
plt.style.use('classic')
```

# Histogram

```python
y=[20,25,30,35,40,45,50]
plt.hist(y,bins=10,color='y')
```

```python
import matplotlib.pyplot as plt

data = [45,50,52,55,60,61,63,70,72,75,80,82,90]

plt.hist(
    data,
    bins=5
    ,
    color='skyblue',
    edgecolor='red'
)

plt.title("Students Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```
