# Merge = Join في سكوال

```python
data1= pd.DataFrame({"Employe":["Emp1","Emp2","Emp3"], "Teams":["Team1","Team2","Team3"]})
data2=pd.DataFrame({"Employe":["Emp1","Emp2","Emp3"], "Hiring_Date":["2010","2020","2023"]})
```

```python
data3=pd.merge(data1,data2)
```

```python
data3
```

```python
data4=pd.DataFrame({"Teams":["Team1","Team2","Team3"],"Salary":[30000,15000,12000]})
```

```python
data5=pd.DataFrame({"Department":["Team1", "Team2", "Team3"], "Newsalary":[25000,20000,15000]})
```

```python
Final_join=pd.merge(data4,data5,left_on="Teams",right_on="Department")
```

```python
Final_join
```
